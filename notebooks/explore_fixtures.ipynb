```python
# Importação dos módulos
import pandas as pd
import requests
import os
from dotenv import load_dotenv
```

```python
# Leitura de variáveis e parâmetros

load_dotenv()

API_KEY = os.getenv('KICK_OFF_API_KEY')

HEADERS = {
    'x-api-key': API_KEY
    }

URL = 'https://app.kickoffapi.com/api/v1/fixtures'

PARAMS = {
    'league': 1,
    'season': 2026
    }
```

```python
# Requisição HTTP

response = requests.get(
    URL,
    headers=HEADERS,
    params=PARAMS,
    timeout=30)

response.raise_for_status()

data = response.json()
```

```python
# Criação do dataframe

df = pd.json_normalize(data["response"])

df.columns.tolist()

df.iloc[0]
```
